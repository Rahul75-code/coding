```python
strs = ["eat","tea","tan","ate","nat","bat"]
```

```python
dic = {}
for i in strs:
    st = ''.join(sorted(i))
    if st in dic:
        dic[st].append(i)
    else:
        dic[st] = [i]
print(dic.values())
```

```python
dic = {}
for i in strs:
    print('i-',i)
    st = ''.join(sorted(i))
    print('st-',st)
    if st in dic:
        dic[st].append(i)
        print('append dic-',dic)
    else:
        dic[st]=[i]
        print('added key dic-',dic)
anagrplist = dic.values()
print(anagrplist)
```
