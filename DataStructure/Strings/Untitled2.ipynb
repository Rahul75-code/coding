```python
s = "   fly me   to   the moon  "
```

```python
s1 = s.strip()
print(s1)
c = 0
for i in range(len(s1)-1,-1,-1):
    if s1[i] ==' ':
        print('break: ', c)
    else :
        c+=1
print(c)
```

```python
strs = ["flower","flow","flight"]
```

```python
r = ''
for i in range(len(strs[0])-1):
    print(i)
    for s in strs:
        print(s)
        if s[i] != strs[0][i]:
            print('break',r)
    r += strs[0][i]       
    print(r)
print(r)
```
