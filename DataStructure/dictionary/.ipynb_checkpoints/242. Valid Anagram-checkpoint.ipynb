```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
         
        if len(s) != len(t):
            return False
        m_s, m_t = {}, {}
        for i in range(len(s)):
            m_s[i] = 1+ m_s.get(i,0)
            m_t[i] = 1+ m_t.get(i,0)
        for c in m_s:
            if m_s[c] != m_t[c]:
                return False
        return True
```
