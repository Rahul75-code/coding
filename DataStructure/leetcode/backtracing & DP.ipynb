wirte the python code to find the sum of number from list a, which equals to b

```python
a = [10, 12, 10, 15, -1, 7, 6, 5, 4, 2, 1, 1, 1]
b = 11

result = []

def dfs(nums, target, index, path):
    if target == 0:
        result.append(path)
        return
    if target < 0 or index >= len(nums):
        return
    
    # Include current number
    dfs(nums, target - nums[index], index + 1, path + [nums[index]])
    
    # Exclude current number
    dfs(nums, target, index + 1, path)

dfs(a, b, 0, [])

print("Subsets that sum to", b, ":")
for subset in result:
    print(subset)

or

from itertools import combinations

a = [10, 12, 10, 15, -1, 7, 6, 5, 4, 2, 1, 1, 1]
b = 11

result = []

# Check all possible subset sizes
for r in range(1, len(a) + 1):
    for combo in combinations(a, r):
        if sum(combo) == b:
            result.append(combo)

print("Subsets that sum to", b, ":")
for subset in result:
    print(subset)

or

a = [10, 12, 10, 15, -1, 7, 6, 5, 4, 2, 1, 1, 1]
b = 11

n = len(a)

# DP table: n+1 rows, b+1 columns
dp = [[False] * (b + 1) for _ in range(n + 1)]

# Base case: sum 0 is always possible (empty subset)
for i in range(n + 1):
    dp[i][0] = True

# Fill DP table
for i in range(1, n + 1):
    for j in range(b + 1):
        # Exclude current element
        dp[i][j] = dp[i-1][j]
        # Include current element if possible
        if j - a[i-1] >= 0:
            dp[i][j] = dp[i][j] or dp[i-1][j - a[i-1]]

# Check if sum b is possible
print("Is sum", b, "possible?", dp[n][b])
```
