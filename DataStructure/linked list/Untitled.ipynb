```python
class Node():
    def __init__(self, data):
        self.data = data
        self.next = None
        print(data)
```

```python

class LinkedList():
    def __init__(self):
        self.head = None
        
    def insert (self, newNode):

        if self.head is None:
            self.head = newNode
        else :
            lastNode = self.head
            while True:
                if lastNode.next is None:
                    break
                lastNode =lastNode.next
            lastNode.next = newNode
            print('New data inserted',newNode.data)
        
    def printList(self):
        if self.head is None:
            print('list is empty')
            return
        currentNode = self.head
        while True:
            print(currentNode.data)
            if currentNode.next is None:
                break
            currentNode = currentNode.next
            
            
    
    
```
