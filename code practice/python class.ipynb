```python
class Employee:
    def __init__(self, employee_id, name):
        self.employee_id = employee_id
        self.name = name
        
    def print_info(self):
        print("Employee ID:", self.employee_id)
        print("Name:", self.name)

# Example usage
```

```python
emp1 = Employee(1, "John Doe")
emp1.print_info()
```

```python
class Employee:
    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id
    def print_info(self):
        print("Employee ID:", self.employee_id)
        print("Name:", self.name)
emp1 = Employee( 'rahul', 12)
emp1.print_info()
```
