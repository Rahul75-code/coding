```python
import urllib
import numpy as np
import ssl
import cv2
```

```python
# Replace 'Your URL' with the IP address provided by the IP Webcam app
url = 'Your URL'

while True:
    imgResp = urllib.urlopen(url)
    imgNp = np.array(bytearray(imgResp.read()), dtype=np.uint8)
    img = cv2.imdecode(imgNp, -1)
    cv2.imshow('Camera Feed', cv2.resize(img, (600, 400)))

    # Press 'q' to exit the loop
    q = cv2.waitKey(1)
    if q == ord("q"):
        break

cv2.destroyAllWindows()
```
