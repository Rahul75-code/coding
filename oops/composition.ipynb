```python
# Composition over inheritance: Instead of making Car inherit from Engine, you inject an Engine object. 
# This makes Car flexible — you can swap in different engine types without changing the Car class.
#Dependency Injection: You pass the dependency (Engine) into Car, which makes testing and extension easier.

# Passing Objects Between Classes:
class Engine:
    def start(self):
        return "Engine started"

class ElectricEngine:
    def start(self):
        return "Electric engine started silently"

class DieselEngine:
    def start(self):
        return "Diesel engine roaring to life"

class Car:
    def __init__(self, engine1):
        self.engine1 = engine1  # Accepting Engine object

    def drive(self):
        print(self.engine1.start())
        print("Car is driving")

# Create an Engine object
engine = Engine()

# Pass Engine object to Car
car = Car(engine)
car.drive()

# Use Electric Engine
car1 = Car(ElectricEngine())
car1.drive()

# Use Diesel Engine
car2 = Car(DieselEngine())
car2.drive()
```

### ETL using composition

```python
class Ingestion:
    def run(self):
        return "Data ingested from ADLS"

class Transformation:
    def run(self, data):
        return f"Transformed {data} with PySpark"

class Storage:
    def run(self, data):
        return f"Stored {data} in Delta Lake"

class Pipeline:
    def __init__(self, ingestion, transformation, storage):
        self.ingestion = ingestion
        self.transformation = transformation
        self.storage = storage

    def execute(self):
        raw = self.ingestion.run()
        transformed = self.transformation.run(raw)
        result = self.storage.run(transformed)
        return result

# Composition in action
pipeline = Pipeline(Ingestion(), Transformation(), Storage())
print(pipeline.execute())
```
