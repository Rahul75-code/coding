```python
# calling a function from another class inside a class
class ClassA:
    def greet(self):
        return "Hello from ClassA!"

class ClassB:
    def call_class_a(self):
        # Create an instance of ClassA
        a = ClassA()
        # Call the greet method from ClassA
        message = a.greet()
        print("ClassB received message:", message)

# Create an instance of ClassB and call its method
b = ClassB()
b.call_class_a()
```

```python
# Inheritance:
class Animal:
    def speak(self):
        return "Animal sound"

class Dog(Animal):  # Dog inherits from Animal
    # pass
    def speak(self):
        return "Woof!"

# Using the inherited method
dog = Dog()
print(dog.speak())  # Output: Woof!
```

```python
# Composition over inheritance: Instead of making Car inherit from Engine, you inject an Engine object. 
# This makes Car flexible — you can swap in different engine types without changing the Car class.
#Dependency Injection: You pass the dependency (Engine) into Car, which makes testing and extension easier.

# Passing Objects Between Classes:
class Engine:
    def start(self):
        return "Engine started"
class ElectricEngine:
    def start(self):
        return "Electric engine started silently"
class DieselEngine:
    def start(self):
        return "Diesel engine roaring to life"
class Car:
    def __init__(self, engine1):
        self.engine1 = engine1  # Accepting Engine object
    def drive(self):
        print(self.engine1.start())
        print("Car is driving")

# Create an Engine object
engine = Engine()
# Pass Engine object to Car
car = Car(engine)
car.drive()
# Use Electric Engine
car1 = Car(ElectricEngine())
car1.drive()
# Use Diesel Engine
car2 = Car(DieselEngine())
car2.drive()
```
