```python
import functools
import logging

# Set up standard logger configuration
logger = logging.getLogger(__name__)
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")

def log_calls(func):
    """A decorator that logs the execution metadata of a function professionally."""
    @functools.wraps(func)  # Fix 1: Preserves the original function's name and docstring
    def wrapper(*args, **kwargs):
        # Fix 2 & 3: Use logging instead of print, and inject dynamic context
        logger.info(f"Calling '{func.__name__}' with args={args}, kwargs={kwargs}")
        
        try:
            result = func(*args, **kwargs)
            logger.info(f"'{func.__name__}' executed successfully.")
            return result
        except Exception as e:
            # Bonus: Logs errors instantly if the function crashes
            logger.error(f"'{func.__name__}' failed with exception: {e}")
            raise e
            
    return wrapper
```

In Python dataclasses, default_factory is a configuration parameter used to safely assign a fresh, brand-new mutable object (like a list, dictionary, or set) to a field every time a new instance of the class is created.
It accepts a callable (a function, class constructor, or lambda expression) and invokes it behind the scenes to generate a unique default value whenever you instantiate the object.

The Problem It Solves: The "Shared Memory" Bug

In standard Python classes or naive dataclass setups, if you try to assign a default mutable value like this:
python

Use code with caution.
Python creates only one list in memory when the class blueprint is first read. Every single object you instantiate points to that exact same list. If one object changes the list, every other object changes too.

```python
# Metaclass Basics

# What will be the output if this code is executed? Why?

class LoggingMeta(type):
    def __new__(mcls, name, bases, namespace):
        print(name)
        return super().__new__(mcls, name, bases, namespace)

class User(metaclass=LoggingMeta):
    pass
```

This code demonstrates how a metaclass works in Python. A metaclass is essentially a "class constructor"—just like a normal class acts as a blueprint for creating objects, a metaclass acts as a blueprint for creating classes.
Here is the exact lifecycle breakdown of what happens behind the scenes:
1. Metaclass Definition (LoggingMeta):
The class LoggingMeta inherits from type. This explicitly marks it as a metaclass. The __new__ method inside it intercepts the creation process of any class that hooks into it.
2. Class Declaration Interception:
When Python reads the class definition line class User(metaclass=LoggingMeta):, it does not wait for you to instantiate the user (you don't need to call User()). The mere act of compiling the class declaration triggers the metaclass machinery.
3. Executing __new__:
Python passes the components of the User class to LoggingMeta.__new__:
	• mcls: The metaclass itself (LoggingMeta)
	• name: The string name of the class being created ("User")
	• bases: A tuple of base classes (empty in this case, ())
	• namespace: A dictionary containing the class variables and methods (empty here, {'__module__': '__main__', '__qualname__': 'User'})
4. The Triggered Actions:
	• Inside __new__, the statement print(name) executes immediately, printing "User" to the console.
	• Finally, super().__new__(...) completes the operation by allocating space in memory and formally registering the new User class so it can be used later in your script.

Why Does It Occur?

In Python, defining a function with the async def keyword means it becomes an asynchronous coroutine function. When you call get_total(), Python does not actually execute the code inside the function or return the integer 12 right away.
Instead, calling get_total() merely creates and returns a coroutine object—essentially a paused package of work waiting to be executed by the event loop.
Because total is holding this raw coroutine object rather than the inner integer value, trying to add them together (total + 1) is mathematically impossible, forcing Python to throw a TypeError.

```python
import asyncio

async def get_total():
    return 12

async def main():
    # FIXED: Added 'await' to execute the coroutine and get its returned integer
    total = await get_total()
    print(total + 1)

asyncio.run(main())
```

```python
# Context Manager Exception Handling

# What will be the output if this code is executed? Why?

class IgnoreErrors:
    def __enter__(self):
        print("start")

    def __exit__(self, exc_type, exc_val, exc_tb):
        print(exc_type.__name__)
        return True

with IgnoreErrors():
    raise RuntimeError("boom")

print("done")
```

This code demonstrates the exact mechanism of exception suppression using a Python Context Manager's __exit__ lifecycle method.
Here is the step-by-step breakdown of the execution flow:
1. Entering the Context (__enter__):
When the with IgnoreErrors(): block begins, Python executes the __enter__() method. This prints "start" to the console.
2. Triggering the Exception:
Inside the body of the with block, the statement raise RuntimeError("boom") executes. This immediately halts the inner block and threatens to crash the program.
3. Intercepting via __exit__:
Before Python lets the exception crash the application, it hands the crash data over to the __exit__() method. The three arguments (exc_type, exc_val, exc_tb) are automatically populated with the error details.
	• exc_type holds the actual exception class (<class 'RuntimeError'>).
	• exc_type.__name__ converts this class object into its human-readable string name, which prints "RuntimeError".
4. The Power of return True:
The final line of the __exit__ method returns the boolean value True. In Python's context manager protocol, returning True acts as a silence command. It tells the Python interpreter: "I have fully handled this error. Swallow it, suppress the traceback completely, and proceed with the rest of the script normally."
5. Continuing Execution:
Because the error was successfully silenced, execution drops down smoothly past the with block to the final print statement, printing "done".

This error occurs because you defined __slots__ inside the User class blueprint.
By default, every standard object instance in Python creates a hidden dynamic dictionary called __dict__ to store its attributes. While this allows you to add any new variables on the fly, dictionaries consume a significant amount of extra RAM.
When you declare __slots__ = ("name",), you explicitly tell Python: "Optimize this class by removing the dynamic __dict__ dictionary entirely. Only allow instances to hold the specific attributes listed here."
Here is the exact step-by-step breakdown of the crash:
1. Valid Assignment: Writing user.name = "Alice" works flawlessly because "name" is explicitly registered inside your __slots__ tuple.
2. Invalid Assignment: When you write user.email = "alice@example.com", Python's lookup manager checks the __slots__ collection for "email".
3. The Crash: Because "email" is not found, and because there is no underlying __dict__ dictionary available to catch arbitrary new fields, Python blocks the operation completely and raises an AttributeError.

• Mechanism: Python resolves method calls from left to right using the C3 Linearization / MRO algorithm.
• Execution: When f.count() is called, Python looks inside Fruit first. Finding nothing, it looks into the first parent class listed, Apple. Apple contains a count() method returning 10, so Python executes it instantly and stops searching. Pear.count() is bypassed entirely.

• Code: fuji = Apple(), crisp = fuji, del fuji
• Mechanism: Python's garbage collector only destroys an object in memory when its reference count hits zero.
• Execution: Assigning crisp = fuji creates a second reference pointer pointing to the same instance. Running del fuji merely removes the variable name fuji; it does not erase the underlying object because crisp is still holding it alive. print(crisp) executes successfully, firing the custom __repr__ method which returns 'instance of Apple'.

• Code: print(gala.__color)
• Mechanism: Any attribute inside a class definition beginning with two leading underscores (like __color) triggers Python's internal Name Mangling engine.
• Execution: To prevent accidental variable clashes in subclasses, Python secretly rewrites the attribute name from __color to _ClassName__color (in this case, _Apple__color). When you try to access the raw name gala.__color from outside the class definition, Python cannot find it and throws a runtime AttributeError.

```python
# Identify the issue in the following code and suggest how to fix it.

from datetime import datetime

def msg_line(file_handle, msg, msg_type, msg_time=datetime.now()):
    # msg_type should be one of A, E, G, H, L, or P.
    if len(msg_type) != 1 or msg_type not in "AEGHLP":
        raise ValueError("msg_type %s is not valid" % msg_type)
    line = "%s (%s): %s\n" % (msg_time, msg_type, msg)
    # May raise an OSError
    file_handle.write(line)
```

The issue in this code is a classic Python anti-pattern: using a mutable/dynamic expression (datetime.now()) as a default parameter value.
In Python, default arguments are evaluated only once when the function is first defined (at compilation/import time), not every time the function is called.
• As a result, msg_time is locked to the exact timestamp of when the script or module first loaded.
• Every subsequent call to msg_line that relies on the default value will log the identical initialization time, completely breaking your timestamp logging utility.

```python
from datetime import datetime

# FIXED: Default value is set to None
def msg_line(file_handle, msg, msg_type, msg_time=None):
    # msg_type should be one of A, E, G, H, L, or P.
    if len(msg_type) != 1 or msg_type not in "AEGHLP":
        raise ValueError("msg_type %s is not valid" % msg_type)
        
    # FIXED: Generate the timestamp dynamically at execution runtime
    if msg_time is None:
        msg_time = datetime.now()
        
    line = "%s (%s): %s\n" % (msg_time, msg_type, msg)
    # May raise an OSError
    file_handle.write(line)
```

An Approach That Will Work: Multiprocessing

To successfully parallelize CPU-bound code, you must use multiprocessing (via Python's built-in multiprocessing or concurrent.futures.ProcessPoolExecutor modules).
Instead of creating threads within a single operating system process, multiprocessing spawns entirely separate Python interpreter processes.
• Bypassing the GIL: Each separate process gets its own distinct memory space and its own personal GIL.
• True Parallel Core Utilization: The operating system can distribute these individual processes across all your available physical CPU cores, achieving true parallel execution.





```python
from concurrent.futures import ProcessPoolExecutor

def heavy_cpu_math_task(data):
    # Imagine high-intensity CPU computations here
    return sum(i * i for i in range(data))

def parallel_execution():
    dataset = [10_000_000, 10_000_000, 10_000_000, 10_000_000]
    
    # ProcessPoolExecutor splits these tasks across separate CPU cores
    with ProcessPoolExecutor() as executor:
        results = list(executor.map(heavy_cpu_math_task, dataset))
    return results
```


When Threading Is Actually Appropriate

Multi-threading is highly effective when your code is I/O-bound (Input/Output-bound) rather than CPU-bound.
An application is I/O-bound when it spends most of its time waiting for external systems or hardware responses, during which the CPU sits idle doing no work.

Key I/O-Bound Scenarios for Threading:

• Network Tasks: Scraping websites, hitting external web APIs, or downloading multiple files simultaneously.
• Database Operations: Waiting for heavy SQL database queries to finish executing and return records.
• File System Operations: Reading or writing large assets onto a slow hard drive array.

```python
from collections import defaultdict

def count_log_line_types(filename):
    """Counts types of log lines in a log format efficiently using streaming."""
    VALID_TYPES = {"TRACE", "DEBUG", "INFO", "WARN", "ERROR", "FATAL"}
    
    # 1. Using a defaultdict removes the need for .get(key, 0) boilerplate
    counts = defaultdict(int)
    
    # 2. 'with' ensures the file is closed automatically, even if an exception occurs
    with open(filename, 'r', encoding='utf-8') as file_in:
        # 3. Iterating directly over the file object streams lines one-by-one (O(1) Memory)
        for line in file_in:
            # Extract the first word safely by splitting on spaces
            log_type = line.split(maxsplit=1)[0]
            parts = line.split(maxsplit=1) 
            if not parts: continue 			# Skip completely blank lines safely 
            log_type = parts[0]
            
            if log_type in VALID_TYPES:
                counts[log_type] += 1
            else:
                raise ValueError(f"Log line {line.strip()} was invalid")
                
    return dict(counts)
```

```python
# Threading
# The following code runs a CPU-intensive calculation both sequentially and with 
# threading. What will be the result? Why?

import threading
import time

def cpu_task():
    total = 0
    for i in range(5000000):
        total += i ** 2
    return total

# Sequential execution
start = time.time()
cpu_task()
cpu_task()
sequential_time = time.time() - start

# Threaded execution
start = time.time()
t1 = threading.Thread(target=cpu_task)
t2 = threading.Thread(target=cpu_task)
t1.start()
t2.start()
t1.join()
t2.join()
threaded_time = time.time() - start

print(f"Sequential: {sequential_time:.2f}s")
print(f"Threaded: {threaded_time:.2f}s")
```

Above code O/P - Sequential: 0.95s Threaded: 1.05s
🔍 Why? (The Mechanics of the CPython GIL)
This counter-intuitive result happens because of the Global Interpreter Lock (GIL) in Python's standard CPython implementation.
1. The Task is Purely CPU-Bound
The cpu_task() function consists entirely of math calculations (i ** 2) and loop operations. This means it requires 100% processor execution power and never pauses to wait for outside factors like a network response, database lookup, or hard drive file stream.
2. The GIL Blocks Parallel Execution
The GIL is a safety lock that ensures only one thread can execute Python bytecode at any given microsecond.
•	In the Sequential run, Core A of your processor completes the first math loop entirely, then completes the second loop.
•	In the Threaded run, you spawn two threads, but they cannot run side-by-side on Core A and Core B. Instead, they must constantly pass the single GIL lock back and forth between each other. They split the processing time on a single CPU core.
3. The Threading Penalty
Because the two threads are constantly fighting for control of the GIL, the operating system is forced to repeatedly stop what it's doing, switch out the context of the CPU registers, and pass the lock over to the competing thread. This overhead (called context switching lock contention) adds processing delays, making the threaded version slightly slower than simply running them back-to-back.


```python
# Improvement:
from multiprocessing import Process
import time

# ... keeping the same cpu_task logic ...

if __name__ == '__main__':
    start = time.time()
    
    # 🌟 Spawning individual processes bypasses the GIL entirely
    p1 = Process(target=cpu_task)
    p2 = Process(target=cpu_task)
    p1.start()
    p2.start()
    p1.join()
    p2.join()
    
    print(f"Multiprocessing Time: {time.time() - start:.2f}s")
    # This will execute roughly twice as fast as the sequential run!
```

If this code is executed to process 1,000 files, it will trigger two severe infrastructure bottlenecks that could crash the script or destabilize the host system: an Operating System crash due to unclosed file limits and an Out-of-Memory error.
What will happen:
The line lines = file_in.readlines() tells Python to swallow the entire contents of the log file into RAM at once, turning every row into an individual string element within a massive list.
The consequence:
Production log files are routinely multi-gigabyte files. If even a handful of these 1,000 files are exceptionally large, reading them into RAM simultaneously will saturate the system's memory cache. 🛠️ The Production-Ready Refactor
To completely bypass memory allocations and guarantee that open file streams are immediately terminated upon completion, rewrite the function to stream the files lazily using a with context manager:

def count_errors(filename):
    count = 0
    # 1. 'with' ensures the file handle is locked down and safely closed immediately
    with open(filename, 'r', encoding='utf-8') as file_in:
        # 2. Iterating directly over file_in reads lines one by one (O(1) Memory footprint)
        for line in file_in:
            if "ERROR" in line:
                count += 1
    return count
🔍 Why the Refactor Works:
•	Zero Leakage: The with statement safely handles file closing the absolute millisecond the loop exits (or crashes), ensuring you never cross the operating system file descriptor limits.
•	Constant Memory footprint: Iterating over file_in directly loads only one single line into memory at a time. The function can scan a 500-Gigabyte file smoothly while consuming just a few kilobytes of RAM.


```python
def count_errors(filename):
    count = 0
    # 1. 'with' ensures the file handle is locked down and safely closed immediately
    with open(filename, 'r', encoding='utf-8') as file_in:
        # 2. Iterating directly over file_in reads lines one by one (O(1) Memory footprint)
        for line in file_in:
            if "ERROR" in line:
                count += 1
    return count
```

```python
# The following code produces an error. 
# Identify the issue and explain why it occurs.

class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

class Dog(Animal):
    def __init__(self, name, breed):
        self.breed = breed

    def speak(self):
        return f"{self.name} barks"

dog = Dog("Buddy", "Labrador")
print(dog.speak())
```

🔍 Why It Occurs
The error occurs because the Dog subclass overrides the parent class's __init__ constructor but fails to initialize the parent class.
Here is the exact step-by-step breakdown of the failure:
1.	Overriding Without Linking: When you write class Dog(Animal):, Dog inherits all behaviors from Animal
. However, by defining a new def __init__(self, name, breed): inside Dog, you are completely overwriting the parent constructor.
2.	Missing Attribute: Inside Dog.__init__, the code only assigns self.breed = breed. It completely ignores the name argument, meaning the variable self.name is never created in memory.
3.	The Crash: When you run dog.speak(), Python looks inside the Dog version of the speak method:


```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

class Dog(Animal):
    def __init__(self, name, breed):
        # 🌟 FIX: Trigger the parent class constructor to set self.name
        super().__init__(name) 
        self.breed = breed

    def speak(self):
        return f"{self.name} barks"

# --- Testing the Fix ---
dog = Dog("Buddy", "Labrador")
print(dog.speak())  # Output: Buddy barks
```
